# 06. Publication Figure 2: Topic-by-Position Selection Heatmap

## Analytical Goal
Examine whether positional primacy is homogeneous across distinct topic domains ($T_1, \dots, T_k$) or if domain familiarity moderates scanning behavior.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/mnt/data/h2_testing_dataset.csv')

# Pivot table: Topic ID vs Displayed Position
pivot_table = df.pivot_table(index='topic_id', columns='displayed_position', values='selected', aggfunc='mean')

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(pivot_table, cmap='YlGnBu', annot=False, cbar_kws={'label': 'Mean Selection Probability'}, ax=ax)

ax.set_title("Figure 2: Attention Allocation Across Topics and Positional Ranks", fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel("Displayed Position Index (1 to 20)", fontsize=11)
ax.set_ylabel("Topic ID", fontsize=11)

plt.tight_layout()
plt.savefig('/mnt/data/github_replication_suite/notebooks/figure2_topic_position_heatmap.png', dpi=300)
plt.show()
print("Figure 2 rendered and exported successfully.")
```
