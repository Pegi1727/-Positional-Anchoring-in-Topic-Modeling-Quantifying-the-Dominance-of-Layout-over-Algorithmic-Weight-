# 10. Publication-Ready Academic Tables Generation (LaTeX & CSV)

## Standard Output
Exports peer-reviewed ready LaTeX and CSV summary tables formatted according to APA/IEEE guidelines for the main manuscript.


```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf

df = pd.read_csv('/mnt/data/h2_testing_dataset.csv')

# Table 1: Descriptive Statistics by Condition
t1 = df.groupby('condition').agg(
    Total_Observations=('selected', 'count'),
    Total_Selections=('selected', 'sum'),
    Selection_Rate=('selected', 'mean'),
    Selection_Std=('selected', 'std'),
    Mean_Model_Weight=('model_weight', 'mean')
).reset_index()

t1_latex = t1.to_latex(index=False, float_format="%.4f", caption="Table 1: Aggregate Selection Performance by Presentation Condition", label="tab:desc_stats")
print("=== TABLE 1: LATEX ===")
print(t1_latex)

with open('/mnt/data/github_replication_suite/notebooks/table1_descriptive.tex', 'w') as f:
    f.write(t1_latex)
t1.to_csv('/mnt/data/github_replication_suite/notebooks/table1_descriptive.csv', index=False)
```

```python
# Table 2: Regression Model Coefficients Table
model = smf.logit("selected ~ displayed_position + model_weight + C(condition)", data=df).fit(disp=False)

summary_table = pd.DataFrame({
    'Coefficient': model.params,
    'Std_Error': model.bse,
    'z_score': model.tvalues,
    'p_value': model.pvalues,
    'Odds_Ratio': np.exp(model.params)
})

t2_latex = summary_table.to_latex(float_format="%.4f", caption="Table 2: Logit Regression Estimates on Human Selection Behavior", label="tab:logit_model")
print("=== TABLE 2: LATEX ===")
print(t2_latex)

with open('/mnt/data/github_replication_suite/notebooks/table2_regression.tex', 'w') as f:
    f.write(t2_latex)
summary_table.to_csv('/mnt/data/github_replication_suite/notebooks/table2_regression.csv')
print(" All tables exported to LaTeX and CSV successfully.")
```
